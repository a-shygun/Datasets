```python
import random

xlist = [random.randint(0,100) for i in range(100)]

def sumlist(*args):
    x = 0
    for i in args:
        x += i
    return x


print(sumlist(*xlist))
```

```python
xlist = [random.randint(0,100) for i in range(100)]
xdict = {}

for i, j in enumerate(xlist):
    xdict[str(i)] = j

def sumdict(**kwargs):
    x = 0
    for i, j in kwargs.items():
        x += j
    return x

print(sumdict(**xdict))
```
