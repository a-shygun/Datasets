```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

df = pd.read_csv('../Data/Insurance.csv')
df.head()
```

```python
df['sex'] = df['sex'].map({'female': 0, 'male': 1})
df['smoker'] = df['smoker'].map({'yes': 0, 'no': 1})
df['region'] = df['region'].map({'southwest': 0, 'southeast': 1, 'northwest': 2, 'northeast': 3})
df.head()
```

```python
x = df.iloc[:, :-1].values
y = np.array(df['charges'].values).reshape(-1,1).ravel()
```

```python
from sklearn.model_selection import   train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
model = RandomForestRegressor(random_state=0, n_estimators=10)
model.fit(x_train, y_train)
print(model.score(x_train, y_train))
print(model.score(x_test, y_test))
```

```python
new_df = pd.DataFrame([{
    'age': int(input('Enter age: ')),
    'sex': input('male[1]/female[0]: '),
    'bmi': float(input('Enter BMI: ')),
    'children': int(input('Enter number of children: ')),
    'smoker': input('smoker[1]/not smoker[0]: '),
    'region': input('southwest[0]/southeast[1]/northwest[2]/northeast[3]: ')
}])
prediction = model.predict(new_df)
print(prediction[0])
```
