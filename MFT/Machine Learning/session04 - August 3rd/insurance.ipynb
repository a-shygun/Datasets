```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import OneHotEncoder, StandardScaler, PolynomialFeatures
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('../data/insurance.csv')
df.head()
```

```python
df['region'].value_counts()
```

```python
X = df.iloc[:, :-1]

cols = ['sex', 'smoker', 'region']
ct = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(drop='first'), cols)
    ],
    remainder='passthrough'
)

X = ct.fit_transform(X)
y = df.iloc[:, -1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

d1_model = LinearRegression()
d1_model.fit(x_train, y_train)

print('1-degree r2 squared score on trained data: ', d1_model.score(x_train, y_train))
print('1-degree r2 squared score on tested data: ', d1_model.score(x_test, y_test))
```

```python
poly2 = PolynomialFeatures(degree=2)
X_poly = poly2.fit_transform(X)
x_train, x_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=0)

d2_model = LinearRegression()
d2_model.fit(x_train, y_train)

print('2-degree r2 squared score on trained data: ', d2_model.score(x_train, y_train))
print('2-degree r2 squared score on tested data: ', d2_model.score(x_test, y_test))
```

```python
poly3 = PolynomialFeatures(degree=3)
X_poly = poly3.fit_transform(X)
x_train, x_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=0)

d3_model = LinearRegression()
d3_model.fit(x_train, y_train)

print('3-degree r2 squared score on trained data: ', d3_model.score(x_train, y_train))
print('3-degree r2 squared score on tested data: ', d3_model.score(x_test, y_test))
```

So what we see is that we want the best and highest r2 score on tested data specifcally. if the r2 score on trained data is more, it means we are seeing signs of overfitting. in 1-degree, both scores are less than satisfactory ( 73% and 79%). by increasing the degree, we are seeing improvements in the r2 score on tested data ( 88% highest of the bunch). if we increase to 3-degree, we are seeing that the score is lowering, showing that the best degree in this scenario is 2.
