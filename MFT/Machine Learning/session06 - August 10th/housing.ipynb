```python
scaler = StandardScaler()
df['longitude'] = scaler.fit_transform(np.array(df['longitude'].values).reshape(-1,1))
df['population'] = scaler.fit_transform(np.array(df['population'].values).reshape(-1,1))
df['households'] = scaler.fit_transform(np.array(df['households'].values).reshape(-1,1))
df["total_bedrooms"] = df["total_bedrooms"].fillna(df["total_bedrooms"].mean(numeric_only=True))

df_encoded = df
df_encoded = pd.get_dummies(df, columns=["ocean_proximity"])
df_encoded.head()
```

```python
model = make_pipeline(
    PolynomialFeatures(degree=3),
    LinearRegression()
)

x = df_encoded.drop(columns=['median_house_value'], axis=1)
y = df[['median_house_value']].values
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=0, test_size=0.2)
model.fit(x_train, y_train)
print(model.score(x_train, y_train))
print(model.score(x_test, y_test))
```
