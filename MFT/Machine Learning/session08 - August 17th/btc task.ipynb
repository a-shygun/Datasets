```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# data = yf.download('BTC-USD', start='2024-01-01', end='2025-04-01')
# data.to_csv('btc.csv')
data = pd.read_csv('./btc.csv')
# data.drop('Unnamed: 0', axis=1, inplace=True)
data = data.drop(index=[0,1])
data.head()
```

```python
df = data[['Price','Close']].copy()

# df['day'] = range(1, len(df) + 1)
# df['Close'] = df['Close'].astype(float)

df['1D'] = df['Close'].shift(1)
df['2D'] = df['Close'].shift(2)
df['3D'] = df['Close'].shift(3)
df['4D'] = df['Close'].shift(4)
df['5D'] = df['Close'].shift(5)
df['6D'] = df['Close'].shift(6)

df.dropna(inplace=True)
df.head()
```

```python
X = df[['1D','2D','3D','4D','5D','6D']]
y = df['Close']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
from sklearn.ensemble import RandomForestRegressor


model = RandomForestRegressor(n_estimators=200, random_state=0)
model.fit(X_train, y_train)

print("Train R²:", model.score(X_train, y_train))
print("Test R²:", model.score(X_test, y_test))
```
