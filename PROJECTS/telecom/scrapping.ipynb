```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import mplcursors
```

```python
x = pd.DataFrame(data=[
    [1, 50, 100000],
    [1, 500, 450000],
    [1, 300, 360000],
    [1, 200, 270000],
    [1, 100, 180000],
    [3, 200, 360000],
    [3, 1500, 1350000],
    [3, 900, 1080000],
    [3, 600, 810000],
    [3, 300, 540000],
    [6, 500, 750000],
    [6, 600, 1026000],
    [6, 3000, 2565000],
    [6, 1800, 2052000],
    [6, 1200, 1539000],
    [12, 2000, 2400000],
    [12, 1200, 1944000],
    [12, 6000, 4860000],
    [12, 3600, 3888000],
    [12, 2400, 2916000],
    [None, 4, 7200],
    [None, 8, 13600],
    [None, 10, 16000],
    [None, 20, 30000],
    [None, 50, 70000],
    [None, 100, 130000],
    [None, 300, 360000],
    ],
    columns=['month', 'size', 'price']
)
x['price'] *= 1.1
x['1gb price'] = x['price'] / x['size']
```

```python
x
```

```python
x.sort_values('1gb price')
```

```python
plt.ticklabel_format(style='plain', axis='y')
sns.scatterplot(data = x, x = x['1gb price'], y = x['price'] , hue='month');
plt.grid()
```

```python
colors_map = {1: 'red', 3: 'green', 6: 'blue', 12: 'purple'}

plt.ticklabel_format(style='plain', axis='y')

for month, color in colors_map.items():
    subset = x[x['month'] == month]
    plt.scatter(subset['1gb price'], subset['price'], 
                c=color, s=subset['size'], label=f'{month} months', alpha=0.7)

plt.legend(title='Months')
plt.show()
```
