```python
import pandas as pd
```

```python
df = pd.read_csv('../Data/Insurance.csv')
```

```python
df['smoker'] = df['smoker'].map({'yes': 1, 'no': 0})
```

```python
df['sex'] = df['sex'].map({'male': 1, 'female': 0})
```

```python
# from sklearn.preprocessing import LabelEncoder
# le = LabelEncoder()
# df['region'] = le.fit_transform(df['region'])
df['region'] = df['region'].map({'northwest': 0, 'northeast': 1, 'southwest': 2, 'southeast':3})
```

```python
df.to_csv('changed_insurance.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = df.drop('charges', axis=1)
y = df['charges']

poly2 = PolynomialFeatures(degree=2)
X_poly = poly2.fit_transform(X)

x_train, x_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=0)

d2_model = LinearRegression()
d2_model.fit(x_train, y_train)

print('2-degree r2 squared score on trained data: ', d2_model.score(x_train, y_train))
print('2-degree r2 squared score on tested data: ', d2_model.score(x_test, y_test))
```
