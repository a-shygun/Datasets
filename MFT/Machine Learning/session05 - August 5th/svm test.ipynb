```python
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.svm import SVR
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../../../DATASETS/Weekly.csv')
x = df[['Year']].values
y = df[['Volume']].values.ravel()

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)

plots = []
```

#  LINEAR REGRESSION - DEGREE 1 - NO SCALING

```python
model = make_pipeline(
    PolynomialFeatures(degree=1),
    LinearRegression()
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])

sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
p1 = plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Linear Regression - Degree 1 - No Scaling"))

plt.show()
```

#  LINEAR REGRESSION - DEGREE 3 - NO SCALING

```python
model = make_pipeline(
    StandardScaler(),
    PolynomialFeatures(degree=3),
    LinearRegression()
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])

sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
p2 = plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Linear Regression - Degree 3 - No Scaling"))

plt.show()
```

#  LINEAR REGRESSION - DEGREE 3 - STANDARD SCALING

```python
model = make_pipeline(
    PolynomialFeatures(degree=3),
    LinearRegression()
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])

sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Linear Regression - Degree 3 - Standard Scaling"))

plt.show()
```

#  SUPPORT VECTOR MACHINE - DEGREE 1 - NO SCALING

```python
model = make_pipeline(
    PolynomialFeatures(degree=1),
    SVR(kernel='rbf', C=100, epsilon=0.1)
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])

sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Support Vector Machines - Degree 1 - No Scaling"))

plt.show()
```

#  SUPPORT VECTOR MACHINE - DEGREE 3 - NO SCALING

```python
model = make_pipeline(
    PolynomialFeatures(degree=3),
    SVR(kernel='rbf', C=100, epsilon=0.1)
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])

sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Support Vector Machines - Degree 3 - No Scaling"))

plt.show()
```

#  SUPPORT VECTOR MACHINE - DEGREE 3 - STANDARD SCALING

```python
model = make_pipeline(
    StandardScaler(), 
    PolynomialFeatures(degree=3),
    SVR(kernel='rbf', C=100, epsilon=0.1)
)
model.fit(x_train, y_train)

print("R2:", model.score(x_test, y_test))
print("MSE:", mean_squared_error(y_test, model.predict(x_test)))

x_val = np.array([[2008]])
y_val_pred = model.predict(x_val)
print("pred:", y_val_pred[0])


sns.scatterplot(x='Year', y='Volume', data=df)
sns.scatterplot(x=x_val.ravel(), y=y_val_pred.ravel(), color='red')

x_train_sorted = np.sort(x_train, axis=0)
y_pred_sorted = model.predict(x_train_sorted)
plt.plot(x_train_sorted, y_pred_sorted, color='black', label='Model Prediction')
plots.append((x_train_sorted, y_pred_sorted, "Support Vector Machines - Degree 3 - Standard Scaling"))

plt.show()
```

```python
sns.scatterplot(x='Year', y='Volume', data=df)
for x, y, label in plots:
    plt.plot(x, y, label=label)
plt.legend()
plt.grid()
plt.show()
```
