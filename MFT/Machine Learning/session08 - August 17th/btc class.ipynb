```python
import yfinance as yf
import pandas as pd
import numpy as np

data = yf.download('BTC-USD', start='2024-01-01', end='2025-04-01')
data.to_csv('btc.csv', index=False)
df = pd.read_csv('./btc.csv')
df = df.drop(0)
df.head()
```

```python
close = np.array(df['Close'].values).reshape(-1,1).ravel()
x = df.iloc[:, 1:].values
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, close, test_size=0.2, random_state=0)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor().fit(x_train, y_train)
print(model.score(x_train, y_train))
print(model.score(x_test, y_test))
```
